> A loss function measures the gap between a model's predictions and the actual target values. Minimising this loss function guides the model to adjust weights and improve accuracy.

2. Why doesn't finding a minimum always mean simply solving $f′(x)=0$ algebraically?

> Solving $f′(x)=0$ algebraically tells us that $x$ is a stationary point, meaning $x$ could be a local minimum, a local maximum, or neither, which is not the same as finding the minimum.

> This points in the direction of steepest decrease, therefore this can be used to point in the direction to shift the parameters in order to lead to a local minimum.

> After changing the parameters, we are at a new point on the loss surface, where the gradient's direction and magnitude may be different. We therefore recalculate it before deciding the next step.

> A smaller step size will require more steps to reach the local minimum but will prevent overshooting and missing this. Taking steps that are proportional to the length of the gradient can be useful as this will mean smaller steps are taken when getting close to the lcoal minimum which will prevent overshooting

## 3. Learn — From derivative to optimisation

Suppose: $$L(w)$$ is a loss function. Lower values are better.

At the current value of $w$: $$\frac{dL}{dw}>0$$ means increasing $w$ increases the loss.

So to make the loss smaller, you should move $w$ in the negative direction.

Conversely: $$\frac{dL}{dW}<0$$ means increasing $w$ initialy decreases the loss.

This gives the update: $$w_{new}=w_{old}-\eta \frac{dL}{dw}$$ where $\eta$ is the **learning rate**

For several parameters: $$\theta_{new}=\theta_{old}-\eta \nabla L(\theta)$$

The 3Blue1Brown lesson describes this as repeatedly stepping in the downhill direction; it also notes that larher steps can move faster but risk overshooting.

### Definitions

- **Parameter**: something the model is allowed to change.
- **Loss/objective**: a function that assigns a numerical value to how well or poorly the current model parameters satisfy the modelling objective. In this case, lower loss means a better fit.
- **Optimisation**: the process of finding parameter values that minimise the loss/objective.

## 6. Learn — Local minima, global minima and covexity

Gradient descent gives us a direction that reduces the fucntion **locally**. This distinction matters.
- A **local minimum** is lower than nearby points
- A **global minimum** is lower than every other point in the domain

Imagine a landscape with several valleys. Starting in one valley and repeatedly walking downhill may take you to its bottom without telling you whether another valley elsewhere is deeper.

A particularly helpful type of function is a **convex** function. A convex function has a bowl-like shape with no misleading local minima away from the global minimum.

This matters because ordinary one-feature linear-regression with mean squared error gives us a particularly well-behaved optimisation problem.

## 8. Learn — Linear regression becomes an optimisation problem

Now connect this directly to ML. For one feature, our model is: $$\hat{y}_i =wx_i+b$$
The model has two paramters:$$w, b$$

For each observation:
- **actual values** = $y_i$
- **predicted value** = $\hat{y}_i$
- **error** = $\hat{y}_i-y_i$

We need one number summarising the errors across all $n$ observations: we'll used the **mean squared error**: $$L(w,b)=\frac{1}{n}\sum_{i=1}^n(wx_i+b-y_i)^2$$

This is a key conceptual shift. Previously you saw: $$f(x,y)$$ as an abstract two-variable function.

Now we have: $$L(w,b)$$ which is also a two-variable function **but its inputs have meaning**:
- $w\rarr$ slope of regression line
- $b\rarr$ intercept
and its output also has meaning:

> $L(w,b)\rarr$ how badly that particular line fits the data

So changing $w$ and $b$ means moving around a **loss surface**.

The gradient is $$\nabla L(w,b)=\begin{bmatrix}\frac{\partial L}{\partial w} \cr \frac{\partial L}{\partial b}\end{bmatrix}$$

and tells us how the loss responds to changes in those two parameters.

## 9. Derive the linear-regression gradients

### Exercise 4.

Start from: $$L(w,b)=\frac{1}{n}\sum_{i=1}^n(wx_i+b-y_i)^2$$

#### A — derivative with respect to $w$

Find: $$\frac{\partial L}{\partial w}$$

> Outer function: $u^2$
>
> Outer function derivative: $2u$
>
> Inner function: $u=(wx_i+b-y_i)$
>
> Inner function derivative: $\frac{du}{dw}=x_i$
>
> $\frac{1}{n}\sum_{i=1}^{n}2(wx_i+b-y_i)\cdot x_i$
>
> $\frac{1}{n}\sum_{i=1}^{n}2x_i(wx_i+b-y_i)$

#### B — derivative with respect to $b$

Find: $$\frac{\partial L}{\partial b}$$

> Outer function: $u^2$
>
> Outer function derivative: $2u$
>
> Inner function: $u=(wx_i+b-y_i)$
>
> Inner function derivative: $\frac{du}{db}=1$
>
> $\frac{1}{n}\sum_{i=1}^{n}2(wx_i+b-y_i)\cdot 1$
>
> $\frac{1}{n}\sum_{i=1}^{n}2(wx_i+b-y_i)$

#### C — construct the gradient

Write: $$\nabla L(w,b)=\begin{bmatrix}\frac{\partial L}{\partial w} \cr \frac{\partial L}{\partial b} \end{bmatrix}$$ using your two expressions.

> $\nabla L(w,b)=\begin{bmatrix}\frac{1}{n}\sum_{i=1}^{n}2x_i(wx_i+b-y_i) \cr \frac{1}{n}\sum_{i=1}^{n}2(wx_i+b-y_i) \end{bmatrix}$

#### D — interpret what you've derived
Without doing any more algebra, answer:

1. What does a positive value of $$ \frac{\partial L}{\partial w} $$ tell you about increasing $w$?

> Increasing $w$ causes the loss to locally increase while $b$ is held fixed

2. Why does the $w$-gradient involve the input values $x_i$?

> Because $L(w,b)$ is calculated from the mean squared error, which is calculated using the error between the predicted value $\hat{y}$ and the actual value $y$ across all values, and since $\hat{y}_i=wx_i+b$ the loss function $L(w,b)$ will contain input values $x_i$. The $w$-gradient specifically involves input values $x_i$ because we are differentiating with respect to $w$ for the $wx_i$ component of $L(w,b)$, $x_i$ becomes a constant.
>
> Essentially, each observations contribution to the $w$-gradient is weighted by its $x_i$.

3. Why does the $b$-gradient behave differently from the $w$-gradient?

> The $w$-gradient weights each residual by $x_i$, while the $b$-gradient sums/averages the residuals without that $x_i$ weighting.

4. If both partial derivatives are known, how would gradient descent update $w$ and $b$?

> $$\begin{bmatrix}w_{i+1} \cr b_{i+1}\end{bmatrix}=\begin{bmatrix}w_i \cr b_i\end{bmatrix}-\eta \begin{bmatrix}\frac{\partial L}{\partial w} \cr \frac{\partial L}{\partial b} \end{bmatrix}$$
>
> $\boxed{w_{i+1}=w_i-\eta \frac{\partial L}{\partial w}}$
>
> $\boxed{b_{i+1}=b_i-\eta \frac{\partial L}{\partial b}}$

## 10. End-of-session check

1. What is the difference between a model parameter and a loss function?

> A model parameter is a variable the model is allowed to change to better predict the correct outcome for a given input, while the loss function is the function that is used to minimise the residual between the predicted and the real result 

2. What is optimisation trying to find?

> The paramater values that make the loss as small as possible

3. Why does the gradient point in the wrong direction for minimisation?

> Because by definition gradient points in the direction of steepest increase, and to minimise the objective we want to move towards the local global minimum

4. In: $$ \theta_{\text{new}} = \theta_{\text{old}} - \eta\nabla L $$ what does each of these mean?
    - $\theta$
    - $\eta$
    - $\nabla L$

> $\theta$ is the model paramters
>
> $\eta$ is the learning rate, which scales the step size in each gradient descent iteration
>
> $\nabla L$ is the gradient of the loss function $L$

5. What is the difference between a local and global minimum?

> A local minimum is lower than nearby points while a global minimum is lower than every other point in the domain

6. Why is a convex loss particularly convenient?

> Because a convex equation only contains one stationary point, therefore gradient descent will neccessarily point to the global minimum and not just a local minimum value

7. In linear regression, why is: $$ L(w,b) $$ a function of two variables?

> Linear regression finds the best fit for a straight line through data, and the equation of a straight line is made up of two parameters: the slope ($w$) and the y-intercept ($b$) - hence two parameters for $L(w,b)$

8. What information does: $$ \frac{\partial L}{\partial w} $$ give us that: $$ \frac{\partial L}{\partial b} $$ does not?

> $\frac{\partial L}{\partial w}$ tells us how the loss locally changes when we alter the **slope $w$** while holding $b$ fixed.
>
> $\frac{\partial L}{\partial b}$ tells us how the locc locally changes when we alter the **intercept $b$** while holding $w$ fixed.
>
> The $w$-gradient contains an $x_i$ weighting because changing a slope affects predictions differently depending on $x_i$, whereas changing the intercept shifts all predictions equally.

9. Explain this entire process in your own words: $$(w,b)\rarr predictions\rarr loss\rarr\nabla L\rarr new~(w,b)$$

> We start with randomised inital values for $w$ and $b$: $$w_0,~~~b_0$$ and we calculate the predicted output values, $\hat{y}$ based on these initial model parameters given a number of inputs for $x$ using: $$\hat{y}_i=w_0x_i+b_0$$
>
> We then calculate the residual (or the error between these prdicted values and the actual values for y): $$\hat{y}_i-y$$
>
> This residual is used to calculate the the gradient of the loss function:
> $$\nabla L(w,b)=\begin{bmatrix}\frac{\partial L}{\partial w} \cr \frac{\partial L}{\partial b}\end{bmatrix}$$
> where:
> $$\frac{\partial L}{\partial w} = \frac{1}{n}\sum_{i=1}^n2x_i(\hat{y}_i-y_i)$$
> and:
> $$\frac{\partial L}{\partial b} = \frac{1}{n}\sum_{i=1}^n2(\hat{y}_i-y_i)$$
> These formulas come from the use of mean squared error as the loss function. 
>
> Thus new parameter values for $w$ and $b$ can be estimated using:
> $$\boxed{w_{new}=w_{old}-\eta \frac{\partial L}{\partial w}}$$
> $$\boxed{b_{new}=b_{old}-\eta \frac{\partial L}{\partial b}}$$
> With a suitable learning rate $\eta$, repeatedly applying these updates can converge toward a minimum. For ordinary linear regression with MSE, the loss is convex, so a minimum reached this way is global.
