## 3. Product rule and chain rule — Learn

### Product rule
The **product rule** is needed when two functions of $x$ are multiplied:
$$\frac{d}{dx}[u(x)v(x)]=u'(x)v(x)+u(x)v'(x)$$
For example: $$x^2(x+3)$$
contains two functions being multiplied.

### Chain rule
The **chain rule** is needed when one function is inside another: $$f(x)=g(h(x))$$
The: $$f'(x)=g'(h(x))h'(x)$$

A useful way of thinking about this is::
> differentiate the **outside**, keep the **inside**, then multiply by the derivative of the **inside**

# 4. Partial derivatives and gradients — Learn

So far out functions have taken one input: $$f(x)$$

Machine-learning loss functions usually depend on **many parameters**.

For example: $$f(x,y)=x^2+3xy+y^2$$

We can ask:

> how does $f$ change as $x$ changes while $y$ is held fixed?

That's the partial derivative: $$\frac{\partial f}{\partial x}$$ When differentiating with respect to $x$, treat **$y$ as a constant**

Likewise: $$\frac{\partial f}{\partial y}$$ treat $x$ as constant.

All partial derivatives can then be collected into one vector called the **gradient**: 
$$\nabla f=\begin{bmatrix} \frac{\partial f}{\partial x} \cr \frac{\partial f}{\partial y} \end{bmatrix}$$
The gradient therefore tells us how the function changes with respect to **each input simultaneously**.

Geometrically, an important rule is:
> $\nabla f$ points in the direction of steepest increase.

That idea will lead directly into gradient descent.
