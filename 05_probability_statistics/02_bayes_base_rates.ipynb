## 2. Bayes Theorem

1. What is a **prior probability**?

> An initial assessment of how likely an event is to happen before you look at any new data or evidence

2. What does the new **evidence** do to that prior?

> New evidence is $E$. The likelihood $P(E\mid H$ is the probability of observing that evidence assuming the hypothesis $H$ is true. Bayes' theorem uses this likelihood together with the prior to update our belief in $H$

3. What is a **posterior probability**?

> The probaility that the hypothesis is true after observing the evidence

4. In your own words, why can the **base rate** of an event matter a lot when interpreting evidence?

> The base rate is the starting probability of the hypothesis before considering the new evidence. Bayes' theorem updates this prior using the likelihood of the observed evidence, so even strong-looking evidence has to be interpreted relative to how common the event was before hand. Additionally, if $E$ is independent of $H$, then the evidence gives no information and the posterior stays equal to the prior.

## 3. Deriving Bayes' Theorem

We can get Bayes' theorem entirely from the conditional-probability formula you already know. There is only **one new rearrangement** here.

Start with: $$P(H\mid E)=\frac{P(H\cap E)}{P(E)}$$

This asks:
    Given that the evidence $E$ occurred, what proportion of those cases also satisfy hypothesis $H$?

Now write conditional probability the other way around: $$P(E\mid H)=\frac{P(E\cap H)}{P(H)}$$

Because intersection is symmetric, $$P(E\cap H)=P(H\cap E)$$

Rearrange the second equation to isolate the intersection: $$P(H\cap E)=P(E\mid H)P(H)$$

Now substitute that into our first equation: $$P(H\mid E)=\frac{P(E\mid H)P(H)}{P(E)}$$

That is **Bayes' Theorem**

### What each term means

For now, I want this mapping to become automatic:

$$ \underbrace{P(H\mid E)}_{\text{posterior}} = \frac{ \underbrace{P(E\mid H)}_{\text{likelihood}} \underbrace{P(H)}_{\text{prior}} }{ \underbrace{P(E)}_{\text{probability of the evidence}} } $$

Notice in particular that: $$P(E\mid H)\neq (P(H\mid E)$$

They answer completely different questions.

For a medical-test example:
- $H$: person has the disease
- $E$: test is positive
- $P(E\mid H)$: probability of a positive test **given that they have the disease**
- $P(H\mid E)$: probability they **have the disease given that the test was positive**

Bayes lets us get from the first kind of probability to the second.

## 4. When $P(E)$ is not given

This is the more important case, because in realistic Bayes problems the overall proabbility of the evidence often has to be constructed.

Suppose:
- $H$: a person has a disease
- $E$: they test positive

and: $$P(H)=0.01$$ $$P(E\mid H)=0.95$$ $$P(E\mid\neg H)=0.05$$

So:
- disease prevalence is 1%
- sensitivity is 95%
- false positive rate is 5%

We want: $$P(H\mid E)$$

The missing piece is: $$P(E)$$

A positive test can happen in **two mutually exclusive ways**: $$H\cap E$$ or: $$\neg H\cap E$$

So: $$P(E)=P(E\mid H)P(H) + P(E\mid\neg H)P(\neg H)$$

Calculate:
1. $P(\neg H)$

> $P(\neg H) = 1 - P(H) = 1 - 0.01 = 0.99$ 

2. $P(E)$

> $P(E)=P(E\mid H)P(H) + P(E\mid\neg H)P(\neg H)$
>
> $P(E)=0.95(0.01) + 0.05(0.99)=0.059$

3. $P(H\mid E)$ using Bayes’ theorem

> $P(H\mid E)=\frac{P(H)P(E\mid H)}{P(E)}$
>
> $P(H\mid E)=\frac{0.01(0.95)}{0.059}=0.161=16.1/%$

4. Interpret the posterior in one sentence.

> Given that a person has tested positive there is approximately a 16.1% probability that they actually have the disease. Even though the test has a sensitivity of 95% (it successfully detects 95% of people that have the disease), a positive test only corresponds to about 16% probability of disease. This is because the **prevalence of the disease is low**, therefore there are vastly more healthy people available to generate false positives.

## 5. Making the base-rate effect concrete

Let's now express the same problem using **10,000 people** instead of probabilities. This often makes Bayes much easier to reason about.

We still have: $$P(H)=0.01$$ $$P(E\mid H)=0.95$$ $$P(E\mid\neg H)=0.05$$

Imagine **10,000 people** are tested.

Work out:
1. How many of the 10,000 people have the disease?

> $Expected~number~of~infected~people=P(infected)Number~of~people~tested$
>
> $0.01(10,000)=100$

2. Of those people, how many test positive?

> $Number~of~true~positives=P(positive~result\mid infected)Number~of~infected~people$
>
> $0.95(100)=95$

3. How many people do not have the disease?

> $Expected~number~of~noninfected~people=Number~of~people~tested-Expected~number~of~infected~people$
>
> $10,000-100=9,900$

4. Of those people, how many nevertheless test positive?

> $Number~of~false~positives=P(psotive~result\mid not~infected)Number~of~noninfected~people$
>
> $0.05(9,900)=495$

5. Therefore, how many positive tests are there altogether?

> $Positive~results=True~positives+False~positives$
>
> $95+495=590$

6. Of everyone who tested positive, what proportion actually has the disease?

> $Probability~of~infected~given~positive~test=\frac{True~positives}{Total~positives}$
>
> $\frac{95}{590}=0.161=16.1\%$

Don't use Bayes' formula for this one. Just reason through the numbers of people. The final answer should, of course, agree with the $16.1%$ we just calculated.

## 6. How changing the base rate changes the posterior

Now we'll change only one thing.

Keep the exact same test:  $$P(E\mid H)=0.95$$ $$P(E\mid\neg H)=0.05$$

But suppose the disease is much more common: $$P(H)=0.10$$

Before calculating anything, answer:
    Do you expect $P(H\mid E)$ to be **higher, lower, or the same** as the previous $16.1%$? Briefly explain why.

> I would expect $P(H\mid E)$ to be higher. This is beacuse the proportion of the infected population is now higher, therefore the number of non-infected individuals is lower, which results in a lower number of false positives since there are fewer non-infected indivudals available to give false positives. Similarly there will be more infected people to contribute to the true positive population. Overall, the proportion of true positives to false positives in the total positive population higher with a higher prevalence.

Then calculate it using Bayes:

$$ P(H\mid E) = \frac{P(E\mid H)P(H)} {P(E\mid H)P(H)+P(E\mid\neg H)P(\neg H)} $$

Show:
1. $P(\neg H)$

> $P(\neg H) = 1-P(H) = 1-0.1 = 0.9$

2. $P(E)$

> $P(E)=P(H)P(E\mid H)+P(\neg H)P(E\mid\neg H)$
>
> $P(E)=0.1(0.95)+0.9(0.05)=0.14$

3. $P(H\mid E)$

> $ P(H\mid E) = \frac{P(E\mid H)P(H)} {P(E\mid H)P(H)+P(E\mid\neg H)P(\neg H)} $
>
> $ P(H\mid E) = \frac{P(E\mid H)P(H)} {P(E)}$
>
> $ P(H\mid E) = \frac{0.95(0.1)}{0.14} = 0.679 = 67.9\%$

4. One sentence comparing the result with the previous 1% base-rate case

> With a 1% base-rate case, the probability of having the disease given a positive test (the positive predictive value, PPV) is 16.1%. Increasing the base rate to 10% results in a PPV of 67.9%. It is important to note that the sensitivity of the test has remained exactly the same; all we have changed is the proportion of infected to healthy people in the population. This follows the prediction made before performing the calculation.

## 7. Verify Bayes with a NumPy simulation

Now we'll do the one small coding task for this session.

Simulate 100,000 people using the original scenario:  $$P(H)=0.01$$ $$P(E\mid H)=0.95$$ $$P(E\mid\neg H)=0.05$$

Use:

````python
import numpy as np

rng = np.random.default_rng(42)
n = 100_000
````

Your job is to create two Boolean arrays:

```
infected
positive
```
where:
- infected[i] says whether person `i` has the disease
- positive[i] says whether person 'i' tests positive

Try to contruct this **without a Python loop over 100,000 people**.

Then calculate from the simulated data:
1. simulated prevalence;
2. simulated sensitivity, $P(E\mid H)$;
3. simulated false-positive rate, P(E\mid\neg H);
4. simulated posterior, $P(H\mid E)$.

For the final one, think in terms of Boolean masks: $$\frac{number~infected~AND~positive}{number~positive}$$

Before you code it, sketch how you think you can generate the `infected` Boolean array using `rng.random(n) and the 1% prevalence.

````python
p_infected = 0.01
p_true_positive = 0.95
p_false_positive = 0.05

infected = rng.random(n) < p_infected

p_positive = np.where(infected, p_true_positive, p_false_positive)
positive = rng.random(n) < p_positive

simulated_prevalence = np.mean(infected)
simulated_sensitivity = np.count_nonzero(infected & positive) / np.count_nonzero(infected)
simulated_false_positive_rate = np.count_nonzero(positive & ~infected) / np.count_nonzero(~infected)
simulated_posterior = np.count_nonzero(infected & positive) / np.count_nonzero(positive)

print('Simulated prevalence:', simulated_prevalence)
print('Simulated sensitivity', simulated_sensitivity)
print('Simulated false positive rate', simulated_false_positive_rate)
print('Simulated posterior', simulated_posterior)
````

1. What is the difference between a prior, a likelihood, and a posterior?

> A prior is the probability of a hypothesis being true, a likelihood is the probability of some evidence being observed in a subset of the population where the prior hypothesis is true, and a posterior is the probability of the hypothesis being true given that some piece of evidence has been observed

2. Write Bayes’ theorem for hypothesis $H$ and evidence $E$.

> $$P(H\mid E)=\frac{P(H)P(E\mid H)}{P(E)}$$

3. Why is $P(E\mid H)$ not the same thing as $P(H\mid E)$?

> $P(H\mid E)$ answers something totally distinct from $P(E\mid H)$. The former describes the probabiliity of H being true given a subset where E is observed while the latter describes the probability of observing E given a subset where H is true

4. If $P(E)$ is not given, how can you calculate it when $H$ and $\neg H$ are the only possibilities?

> $P(E)=P(H)P(E\mid H) + P(\neg H)P(E\mid\neg H)$
