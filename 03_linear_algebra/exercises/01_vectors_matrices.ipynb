> element-wise takes two matrices of the same dimension or are broadcast-compatible and multiplies elements by their matching positions. matrix multiplication calculates the dot product of rows from the first matrix and columns from the second

```python
import math
import numpy as np
from numpy import linalg as LA

a = np.array([2, 1])
b = np.array([-1, 3])

transformation = np.array([
    [2, 1],
    [0, 1]
])
```

### Part A — vector operations

```python
# a + b — output should be a vector;
c = a + b
# c.shape = (2,)

# 3 * a — output should be a vector;
d = 3 * a
# d.shape = (2,)

# the dot product of a and b — output should be one scalar;
dot_product_np = np.dot(a, b)
# scalar value

# the norm of each vector — each output should be one scalar;
norm_a = LA.norm(a)
norm_b = LA.norm(b)
# both scalar values

# the Euclidean distance between the vectors — output should be one scalar.
distance_between = LA.norm(a - b)
```

### Part B — verify the dot product

```python
dot_product_manual = 0
for i in range(len(a)):
  dot_product_manual += a[i] * b[i]

print(f'Dot products match: {dot_product_np == dot_product_manual}')
```

The dot product multiplies the x components together and adds this to the product of the y components. It tells you how much two vectors point in the same direction

### Part C — matrix-vector multiplication

```python
# input: (2, 2) mult (2,) -> (2,) output
transform_a = transformation @ a
transform_b = transformation @ b
```

(x, y) -> (2x + y, y)

The transformation leaves the y-coordinate unchanged while scaling the original x-coordinate by 2 and adding the y-coordinate to it. It combines horizontal scaling with a horizontal shear
