### Determinant
The factor by which a linear transformation changes any area is called the *determinant* of that transformation. The sign of the determinant denotes whether the transformation is orientation preserving (positive preserves, negative flips). A determinant of zero denotes a transformation that collapses the output space to a lower dimension.

### Invertibility
If the determinant of a linear transformation is non-zero, an inverse of this matrix exists and the matrix is said the be invertible. This is because all output vectors of the transformation can be mapped back to the input vectors by using an inverse of the matrix transformation. In the case when det(A) = 0, the transformation collapses input values to a lower dimension, meaning multiple inputs will map to the same output. As there is no function that can map these outputs back to their multiple original inputs, the inverse transformation does not exist.

A square invertible matrix gives a unique solution to Ax = b because:

- A<sup>-1</sup>•A•x = A<sup>-1</sup>•b

and since: A<sup>-1</sup>•A = I (where I = identity matrix):

- x = A<sup>-1</sup>•b 

i.e., A<sup>-1</sup>•b = the unique solution to the set of equations.

### Singular matrices
A square matrix with a determinant equal to zero, meaning it has no inverse.

Each determinant should be one scalar.

#### A:

- det_A = 2 * 1 - 1 * 1 = 1
- det_A ≠ 0; therefore A is invertible
- A columns span all of R<sup>2</sup> as columns are linearly independent
- A has rank 2

#### B:

- det_B = 1 * 4 - 2 * 2 = 0
- det_B = 0; therefore B is not invertible

#### C:

- det_C = 0 * 0 - (-1) * 1 = 1

```python
# Verify

det_A = np.linalg.det(A)
rank_A = np.linalg.matrix_rank(A)
print(f'det(A): {det_A}; A has rank: {rank_A}')
print('Matrix A has no inverse:', np.isclose(det_A, 0))

det_B = np.linalg.det(B)
rank_B = np.linalg.matrix_rank(B)
print(f'det(B): {det_B}; B has rank: {rank_B}')
print('Matrix B has no inverse:', np.isclose(det_B, 0))

det_C = np.linalg.det(C)
rank_C = np.linalg.matrix_rank(C)
print(f'det(C): {det_C}; C has rank: {rank_C}')
print('Matrix C has no inverse:', np.isclose(det_C, 0))
```

### Part C — calculate and verify an inverse

Choose one matrix you predicted to be invertible.

1. Calculate its inverse with np.linalg.inv().
2. Print the inverse and its shape.
3. Multiply the matrix by its inverse in both orders.
4. Compare each product with a 2 × 2 identity matrix.

```python
inverse_A = np.linalg.inv(A)
print(f'Inverse of A:\n{inverse_A};\nShape: {inverse_A.shape}')

x = A @ inverse_A
y = inverse_A @ A

identity = np.eye(2)

inverse_verified = (
    np.allclose(x, identity)
    and np.allclose(y, identity)
)

print('Inverse verified:', inverse_verified)

print('A @ inverse_A has shape:', x.shape)
print('inverse_A @ A has shape:', y.shape)

```

## 4. Close

1. What does the magnitude of a determinant represent geometrically?

> The scale factor applied by the transformation to the n-dimensional volume.

2. What does a negative determinant indicate?

> The orientation of the basis is reversed after applying the transformation

3. Why does determinant zero prevent a matrix from having an inverse?

> Determinant zero indicates a collapse in vector space. This collapse in dimensionality cannot be reversed by a linear transformation, therefore an inverse transformation cannot exist for a transformation with a determinant of zero 

4. How are determinant, rank and a unique solution related for a square matrix?

> For a square `n x n` matrix, a non-zero determinant, rank `n` and invertibility are equivalent. In that case, `A @ x = b` has one unique solution for every compatible `b`. If the determinant is zero, the rank is below `n`, and a system can either have no solution or infinitely many solutions, but not one unique solution

5. What is the difference between finding an inverse and solving one linear system?

> Finding the inverse constructs a matrix that reverses A for every possible input. Solving one linear system finds the particular vector `x` satisfying `A @ x = b` for one chosen `b`
