Define:

### Linear combination: 
a vector V is called a linear combination of vectors v<sub>1</sub>, v<sub>2</sub>, ..., v<sub>r</sub> if V can be expressed as:

> V = k<sub>1</sub>v<sub>1</sub> + k<sub>2</sub>v<sub>2</sub> + ... + k<sub>r</sub>v<sub>r</sub> 

example:

> np.array([5, 1]) = np.array([2, 2]) + np.array([3, -1])

Linear combinations are the building blocks for vectors in vector spaces

### Span:
the set of all possible vectors that can be reached by making linear combinations of a given group of vectors. If the span covers the all possible values in the space, the starting vectors are called a basis for that space.

example:

In two dimensions, i (1, 0) and j = (0, 1) can be used to express any vector np.array([x, y]):

>  (x, y) = x•i + y•j

### Basis:
The set of linearly independent vectors that span that space

### Linear dependence:
A set of vectors are linearly dependent if one of the vectors can be expressed as a linear combination of the others:

> u = a•v + b•w

for some scalar values a and b. This means the vector already lies in the span of the other vectors

**Correction: A set of vectors is linearly dependent when there is a linear combination equal to the zero vector using coefficients that are not all zero.**

### Linear independence:

If each vector adds another dimension to the span, they are said to be linearly independent:

> u ≠ a•v + b•w

**Correction: A set is linearly independent when the only linear combination equal to the zero vector uses all-zero coefficients.**

## 3. Practical exercise

```python
import numpy as np

u = np.array([1, 2])
v = np.array([2, 4])
w = np.array([-1, 1])
```

### Part A - linear combinations

```python
print(2 * u - v)
# prediction: [0, 0]

print(u + 2 * w)
# prediction: [-1, 4]

print(3 * u - w)
# prediction: [4, 5]
```

A linear combination is the expression of a vector as the addition of other vectors multiplied by scalars

### Part B — dependence and independence
Create two matrices whose columns are:

- Matrix A: u and v
- Matrix B: u and w

```python
A = np.column_stack((u, v))

# print the matrix;
print('Matrix A:\n', A)
print('A shape:', A.shape)

# calculate its rank using np.linalg.matrix_rank();
A_rank = np.linalg.matrix_rank(A)

# A_rank = 1
print('A rank:', A_rank)
```

- decide whether its two column vectors are linearly independent;

> Column vectors are linearly dependent: [1, 2] = 0.5 * [2, 4]

- decide whether the columns span all of two-dimensional space.

> As the column vectors are linearly dependent, they occur on the exact same vector line. This means they do not span all of two-dimensional since any linear combination of the two vector will occur on the line k * [1, 2], therefore only spanning 1 dimension


```python
B = np.column_stack((u, w))

# print the matrix;
print('Matrix B:\n', B)
print('B shape:', B.shape)

# calculate its rank using np.linalg.matrix_rank();
B_rank = np.linalg.matrix_rank(B)

# B_rank = 2
print('B rank:', B_rank)
```

- decide whether its two column vectors are linearly independent;

> Column vectors are linearly independent: [1, 2] ≠ k * [-1, 1]

- decide whether the columns span all of two-dimensional space.

> Since the vectors are linearly independent, any point in two-dimensional space can be achieved by some combination of the two column vectors


### Part C — representing a target vector

Use:

```python
target = np.array([3, 3])
```

For the matrix formed from u and w, find coefficients satisfying:

`coefficient₁ × u + coefficient₂ × w = target`


Solution:

On paper calculation:
> - coefficient₁ - coefficient₂ = 3
> - 2 * coefficient₁ + coefficient₂ = 3

&rarr; coefficient₁ = 2; coefficient₂ = -1

Code solution:

```python
coefficients = np.linalg.solve(B, target)

print('Coefficients:', coefficients)
print('Shape:', coefficients.shape)

# Verify:
reconstructed_target = B @ coefficients

print('Reconstructed target:', reconstructed_target)
print('Matches target:', np.allclose(reconstructed_target, target))
```

### Part E — basis

1. Do u and v form a basis for two-dimensional space?

> No because they are not linearly independent vectors and therefore sit on the same one-dimensional line

2. Do u and w form a basis for two-dimensional space?

> Yes as they are linearly independent vectors and therefore can be linearly combined to make up any point in two-dimensional space 

3. What two requirements must a pair of two-dimensional vectors meet to form a basis for two-dimensional space?

> They must be non-zero and linearly independent vectors

4. How does rank help answer that question?

> If a matrix made up of vector columns has rank 2 then the vectors are linearly independent and form a basis for two-dimensional space

> More generally, a set of vectors form a basis for an n-dimensional space when they are independent and span that space. Consequently, the matrix whose columns are those vectors has rank n

## 4. Close

- What is the difference between span and basis?

> Span is the set of all linear combinations made up by a set of vectors (every point in space that can be reached by any combination of a set of vectors). A basis is a linearly independent set of vectors that spans the space, without any redundant vectors

- How did you recognise that two vectors were dependent?

> If they satisfy the condition: v<sub>1</sub> = k•v<sub>2</sub>; for some scalar k

- What does rank tell you about a matrix?

> Rank tells you the maximum number of linearly independent columns of a matrix, which tells us the dimension of the vector space spanned by its columns
